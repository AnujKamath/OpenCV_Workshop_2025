## 5. Lists, Tuples, and Dictionaries (In Detail)


### Lists
- Lists are ordered, mutable collections of items.
- Defined using square brackets `[]`.

```python
fruits = ["apple", "banana", "cherry"]
print(fruits[0])  # Indexing
fruits.append("mango")  # Add item
print(fruits)
```

### Tuples
- Tuples are ordered, immutable collections.
- Defined using parentheses `()`.

```python
colors = ("red", "green", "blue")
print(colors[1])
```

### Dictionaries
- Dictionaries store data in key-value pairs.
- Defined using curly braces `{}`.

```python
student = {"name": "Alice", "age": 21}
print(student["name"])  # Access by key
student["grade"] = "A"  # Add new key-value
print(student)
```


## 6. Indexing, Looping, and Operations (In Detail)


### Indexing in Lists and Tuples
```python
fruits = ["apple", "banana", "cherry"]
print(fruits[0])     # First element
print(fruits[-1])    # Last element
```

### Looping through Lists
```python
for fruit in fruits:
    print(fruit)
```

### List Operations
```python
numbers = [1, 2, 3]
print(numbers + [4, 5])  # Concatenation
print(numbers * 2)       # Repeat
```

### Looping through Dictionary
```python
student = {"name": "Alice", "age": 21}

for key, value in student.items():
    print(key, ":", value)
```
